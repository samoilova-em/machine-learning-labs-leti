## Описание графика PairGrid

На изображении представлена матрица попарных графиков для пяти признаков (Var1–Var5) с выделением трёх классов (Class1, Class2, Class3). По диагонали расположены графики ядерной оценки плотности распределения каждого признака, выше диагонали — диаграммы рассеяния, ниже диагонали — двумерная ядерная оценка плотности.

Анализ графика показывает, что классы наиболее чётко разделяются в проекциях Var2–Var3 и Var1–Var2, где наблюдаются три отдельных кластера с минимальным перекрытием. Признак Var5 также демонстрирует хорошую разделимость, особенно для Class3. В то же время признак Var4 характеризуется значительным перекрытием всех трёх классов, что указывает на его низкую информативность для задач классификации.

Между признаками выявлены корреляционные зависимости: Var1 и Var4 демонстрируют отрицательную корреляцию, Var4 и Var5 — положительную корреляцию, что проявляется в вытянутой форме контуров плотности. Признаки Var2 и Var3 являются наиболее независимыми.

Распределения признаков имеют различную форму: Var1 показывает бимодальное распределение с разделёнными пиками для разных классов, Var5 имеет явно выраженный отдельный пик для Class3. Данные в целом не являются линейно разделимыми, что предполагает необходимость применения нелинейных методов классификации. Наиболее перспективными для построения моделей представляются признаки Var2, Var3 и Var5, тогда как Var4 может быть исключён из рассмотрения как малoинформативный или сильно коррелирующий с другими признаками.

```python
#2
#2.1, 2.2
import numpy as np 
X = np.loadtxt('lab1_var5.csv', delimiter=',', skiprows=1, usecols=range(1, 6))
y = np.loadtxt('lab1_var5.csv', delimiter=',', skiprows=1, usecols=6)
print(X[:10])
```

```python
#2.3
def describe_np(X):
    for i in range(X.shape[1]):
        column = X[:, i]
        count = np.count_nonzero(column)
        mean = np.mean(column)
        std_unbiased = np.std(column, ddof=1)
        min_val = np.min(column)
        q25 = np.percentile(column, 25)
        median = np.median(column)
        q75 = np.percentile(column, 75)
        max_val = np.max(column)

        print(f"\nVar{i+1}")
        print(f"count: {count}")
        print(f"mean:  {mean:.4f}") 
        print(f"std:   {std_unbiased:.4f}")
        print(f"min:   {min_val:.4f}")
        print(f"25%:   {q25:.4f}")
        print(f"50%:   {median:.4f}")
        print(f"75%:   {q75:.4f}")
        print(f"max:   {max_val:.4f}")

describe_np(X)
```

```python
#3
#3.1
from sklearn.preprocessing import LabelEncoder, OneHotEncoder

labels = df_new['Label'].values
labels_encoder = LabelEncoder()
one_hot_encoder = OneHotEncoder(sparse_output=False)
res_le = labels_encoder.fit_transform(labels)
res_ohe = one_hot_encoder.fit_transform(labels.reshape(-1, 1))

print("До кодирования: ", labels[:10])
print("Кодирование с помощью LabelEncoder: ", res_le[:10])
print("Кодирование с помощью OneHotEncoder:")
print(res_ohe[:10])


```

Для LabelEncoder был создан объект соответствующего класса и вызван метод fit_transform. В результате имена классов были закодированы одним числом (0, 1, 2 соответственно порядку классов). Размерность данных не изменилась.

При создании объекта класса OneHotEncoder параметр sparse_output=False был инициализирован, так как по умолчанию данный класс возвращает разреженную матрицу (sparse matrix), что для данной задачи излишне: всего три категории. При вызове метода fit_transform параметры были переданы в формате двумерного массива размера (n × 1) с помощью метода reshape(-1, 1) — каждый признак должен быть представлен в виде отдельного столбца таблицы. Каждая категория кодируется нулями и единицей: вектор содержит 0 везде, кроме позиции соответствующего класса. Размерность увеличилась до N × K, где K — количество классов.

```python
#3.3
from sklearn.preprocessing import StandardScaler, MinMaxScaler, MaxAbsScaler, RobustScaler
import matplotlib.pyplot as plt

var_columns = df_new.columns[1:6]
X = df_new[var_columns].values 

scalers = {
    'StandardScaler': StandardScaler(),
    'MinMaxScaler': MinMaxScaler(),
    'MaxAbsScaler': MaxAbsScaler(),
    'RobustScaler': RobustScaler()
}

fig, axes = plt.subplots(len(scalers),len(var_columns), figsize=(15, 10))

scaled_data = {}
for name, scaler in scalers.items():
    scaled_data[name] = scaler.fit_transform(X) 

for i, method in enumerate(scalers.keys()):
    for j in range(len(var_columns)): 
        ax = axes[i, j]
        ax.hist(scaled_data[method][:, j], bins=10)
        ax.set_xlabel(var_columns[j])

       
        if j == 0:
            ax.set_ylabel(method)

plt.show()
```
